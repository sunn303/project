```python
import numpy as np
from pathlib import Path

MODEL_DIR = Path(
    "/content/drive/MyDrive/VoiceModel_Output/network_weights_256"
)

print("Loading trained model...")

# Your actual filenames
required_files = [
    "W1.txt",
    "b1.txt",
    "W2.txt",
    "b2.txt",
    "scaler_mean_60.txt",
    "scaler_scale_60.txt"
]

# Check files
for filename in required_files:
    file_path = MODEL_DIR / filename

    if not file_path.exists():
        raise FileNotFoundError(
            f"\nMissing model file:\n{file_path}"
        )

# ------------------------------------------------
# Load neural network weights
# ------------------------------------------------

W1 = np.loadtxt(
    MODEL_DIR / "W1.txt",
    dtype=np.float32
).reshape(60, 256)

b1 = np.loadtxt(
    MODEL_DIR / "b1.txt",
    dtype=np.float32
).reshape(256)

W2 = np.loadtxt(
    MODEL_DIR / "W2.txt",
    dtype=np.float32
).reshape(256, 1)

b2 = float(
    np.asarray(
        np.loadtxt(
            MODEL_DIR / "b2.txt",
            dtype=np.float32
        )
    ).reshape(-1)[0]
)

# ------------------------------------------------
# Load scaler
# ------------------------------------------------

scaler_mean = np.loadtxt(
    MODEL_DIR / "scaler_mean_60.txt",
    dtype=np.float32
).reshape(60)

scaler_scale = np.loadtxt(
    MODEL_DIR / "scaler_scale_60.txt",
    dtype=np.float32
).reshape(60)

# ------------------------------------------------
# Verify
# ------------------------------------------------

print("\nModel loaded successfully.")

print("\nNeural Network:")
print("W1 shape :", W1.shape)
print("b1 shape :", b1.shape)
print("W2 shape :", W2.shape)
print("b2      :", b2)

print("\nScaler:")
print("Mean shape  :", scaler_mean.shape)
print("Scale shape :", scaler_scale.shape)
```

```python
import librosa
import numpy as np

SAMPLE_RATE = 16000
NUM_MFCC = 30
N_MELS = 32
N_FFT = 512

FRAME_LENGTH = int(0.030 * SAMPLE_RATE)
FRAME_SHIFT = int(0.010 * SAMPLE_RATE)


def extract_features(audio_file):

    print("\nReading audio:")
    print(audio_file)

    audio, sr = librosa.load(
        audio_file,
        sr=SAMPLE_RATE,
        mono=True
    )

    print("Sample rate :", sr)
    print("Samples     :", len(audio))
    print("Duration    :", round(len(audio) / sr, 3), "seconds")

    # Remove DC
    audio = audio - np.mean(audio)

    # Peak normalization
    peak = np.max(np.abs(audio))

    if peak > 0:
        audio = audio / peak

    # MFCC
    mfcc = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=NUM_MFCC,
        n_fft=N_FFT,
        hop_length=FRAME_SHIFT,
        win_length=FRAME_LENGTH,
        window="hamming",
        n_mels=N_MELS,
        fmin=10,
        fmax=sr // 2,
        center=False,
        dct_type=2,
        norm="ortho"
    )

    # 30 mean coefficients
    mfcc_mean = np.mean(mfcc, axis=1)

    # 30 standard deviations
    mfcc_std = np.std(mfcc, axis=1)

    # 30 + 30 = 60 features
    features = np.concatenate([
        mfcc_mean,
        mfcc_std
    ]).astype(np.float32)

    print("\nFeature extraction complete.")
    print("MFCC shape :", mfcc.shape)
    print("Feature count :", len(features))

    return features
```

```python
def sigmoid(x):
    x = np.clip(x, -50, 50)
    return 1.0 / (1.0 + np.exp(-x))


def predict_voice(features):

    # 1. Standardize using the SAME scaler from training
    x_scaled = (
        features - scaler_mean
    ) / scaler_scale

    # 2. Dense: 60 → 256
    z1 = np.dot(
        x_scaled,
        W1
    ) + b1

    # 3. ReLU
    hidden = np.maximum(
        0,
        z1
    )

    # 4. Dense: 256 → 1
    z2 = (
        np.dot(
            hidden,
            W2
        )[0]
        + b2
    )

    # 5. Sigmoid
    p_patho = sigmoid(z2)

    p_healthy = 1.0 - p_patho

    # Classification
    if p_patho >= 0.5:
        prediction = "PATHO"
    else:
        prediction = "HEALTHY"

    return prediction, p_healthy, p_patho


print("Prediction function loaded successfully.")
```

```python
from google.colab import files

print("==============================================")
print("       VOICE DISORDER CLASSIFICATION")
print("==============================================")

print("\nUpload a WAV file:")

uploaded = files.upload()

if len(uploaded) == 0:
    raise RuntimeError("No audio file was uploaded.")

audio_file = next(iter(uploaded.keys()))

print("\nSelected file:")
print(audio_file)

# Extract 60 features
features = extract_features(audio_file)

# Run neural network
prediction, p_healthy, p_patho = predict_voice(
    features
)

# ------------------------------------------------
# Display result
# ------------------------------------------------

print("\n")
print("=" * 55)
print("                 RESULT")
print("=" * 55)

print(f"\nPrediction : {prediction}")

print(
    f"P(HEALTHY): {p_healthy * 100:.2f}%"
)

print(
    f"P(PATHO)  : {p_patho * 100:.2f}%"
)

print("\n" + "=" * 55)
```
